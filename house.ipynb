## Preprocessing

```python
from sklearn.preprocessing import  StandardScaler
from sklearn.model_selection import train_test_split
```

```python
features = ["bedrooms", "bathrooms", "sqft_living", "sqft_above","sqft_basement","sqft_lot", "sqft_lot15","sqft_living15","view","floors","condition"]

def fix_skewness(dataset):
    global features

    # Fixing Skewness in data columns

    for feature in features:
        dataset[feature] = np.log(dataset[feature] + 1)


    return dataset

```

```python
df = fix_skewness(df)
```

## Splitting The Data and Applying cross validation and finding out the best model

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import StandardScaler
```

```python
kfold = KFold(n_splits=5)

```

```python
# Using Simple Linear Regression
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline




lr = LinearRegression()

pipeline = make_pipeline(StandardScaler(), lr)

cross_val_score(pipeline,df[features],df.price,cv=kfold, scoring="neg_root_mean_squared_error")
```

```python
# Using Ridge Regression
from sklearn.linear_model import Ridge



alpha_values = [0.01, 0.1, 1, 10, 100, 1000, 10000]
for alpha_value in alpha_values:

    rg = Ridge(alpha=alpha_value)

    pipeline = make_pipeline(StandardScaler(),rg)

    print(f"With Alpha value {alpha_value}, the scores for each fold are:\n\n")
    print(cross_val_score(pipeline,df[features], df.price, cv=kfold, scoring="neg_root_mean_squared_error"),"\n\n")

```

```python
# using random forrest classifier
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=100)

print(cross_val_score(rf,df[features], df.price, cv=kfold,scoring="neg_root_mean_squared_error"))
```

## RandomForrestRegressor performs wayyyy better. Now we will try to find hyperparameters

```python
from sklearn.model_selection import GridSearchCV

rf = RandomForestRegressor()

param_grid = {
    "n_estimators" : [100,150,200,250,300],
    "max_depth" : [2,3,4]
}


search = GridSearchCV(param_grid=param_grid,estimator=rf,scoring="neg_root_mean_squared_error", cv=kfold).fit(df[features],df.price)
```

```python
search.best_params_
```

```python
#  Trying cross validation score with the best estimator found in hyperparameter tuning
```

```python
rf = RandomForestRegressor(n_estimators=200, max_depth=4)

print(cross_val_score(rf,df[features], df.price, cv=kfold,scoring="neg_root_mean_squared_error"))
```
